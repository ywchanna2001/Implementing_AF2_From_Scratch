# Axis-Wise Operations

```python
# Shape(4,3)
A = torch.arange(12).reshape(4,3)
```

```python
# Summing up the rows
A.sum(dim=0)
```

```python
# Summing up the columns
A.sum(dim=1)
```

```python
# Summing up all the elements
A.sum(dim=None)
```

```python
torch.argmax(A, dim=1)
```

```python
torch.argmax(A, dim=0)
```

```python
A = A.float()
torch.mean(A, dim=1)
```

```python
torch.std(A, dim=1)
```

```python
torch.linalg.vector_norm(A, dim=1)
```
